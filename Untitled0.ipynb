```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/superstore_data.csv'
df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
df.groupby('Education')['Income'].mean().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
df.groupby('Education')['Income'].mean().plot(kind='bar', color='teal')
plt.title('Average Income by Education Level')
plt.ylabel('Income')
plt.show()
```

```python
df.groupby('Marital_Status')['MntWines'].mean().sort_values(ascending=False).plot(kind='bar', color='orange')
plt.title('Average Wine Spending by Marital Status')
plt.show()
```
